# Blue Nike Hat Predictor AI

```python
import pandas as pd
from pathlib import Path
from datetime import datetime

# Create a dataset to collect real observations.
FILE = Path("blue_nike_hat_data.csv")

COLUMNS = [
    "weather",
    "hat_available",
    "activity",
    "outfit_matches",
    "mood",
    "nike_preference",
    "hat_habit",
    "wore_blue_nike_hat"
]

if FILE.exists():
    data = pd.read_csv(FILE)
else:
    data = pd.DataFrame(columns=COLUMNS)


def ask_choice(question, options):
    print("\n" + question)

    for i, option in enumerate(options, start=1):
        print(f"{i}. {option}")

    while True:
        try:
            choice = int(input("Choose a number: "))

            if 1 <= choice <= len(options):
                return options[choice - 1]

            print("Please select a listed option.")

        except ValueError:
            print("Enter a number.")


def ask_rating(question, minimum=1, maximum=5):
    while True:
        try:
            value = int(input(f"{question} ({minimum}-{maximum}): "))

            if minimum <= value <= maximum:
                return value

            print(f"Enter a number from {minimum} to {maximum}.")

        except ValueError:
            print("Enter a whole number.")


def collect_observation():
    print("\n=== BLUE NIKE HAT PREDICTOR ===")
    print("Record the circumstances and the actual outcome.")

    observation = {
        "weather": ask_choice(
            "What is the weather?",
            ["Sunny", "Cloudy", "Rainy", "Cold", "Other"]
        ),

        "hat_available": ask_choice(
            "Was a blue Nike hat available to wear?",
            ["Yes", "No"]
        ),

        "activity": ask_choice(
            "What were you doing?",
            ["Football", "Gym", "Going out",
             "Errands", "Travelling", "Other"]
        ),

        "outfit_matches": ask_choice(
            "Did your outfit match the hat?",
            ["Yes", "No", "Not applicable"]
        ),

        "mood": ask_rating(
            "How strongly did your mood influence your choice?",
            1, 5
        ),

        "nike_preference": 4,

        "hat_habit": ask_choice(
            "How often did you wear hats around this period?",
            ["Never", "Rarely", "Sometimes",
             "Often", "Almost always"]
        )
    }

    outcome = ask_choice(
        "Did you actually wear a blue Nike hat?",
        ["Yes", "No"]
    )

    observation["wore_blue_nike_hat"] = (
        1 if outcome == "Yes" else 0
    )

    global data

    data = pd.concat(
        [data, pd.DataFrame([observation])],
        ignore_index=True
    )

    data.to_csv(FILE, index=False)

    print("\nObservation saved!")
    print("Total observations:", len(data))

    print("\nLatest observation:")
    print(pd.DataFrame([observation]).to_string(index=False))


collect_observation()
```

```python
import pandas as pd
from pathlib import Path
from datetime import datetime

FILE = Path("blue_nike_hat_data.csv")

COLUMNS = [
    "date",
    "weather",
    "hat_available",
    "activity",
    "outfit_matches",
    "mood",
    "nike_preference",
    "wore_blue_nike_hat"
]

# Load the existing dataset or create a new one.
if FILE.exists():
    data = pd.read_csv(FILE)

    # Add the date column if your earlier dataset lacks it.
    if "date" not in data.columns:
        data.insert(0, "date", None)

    for column in COLUMNS:
        if column not in data.columns:
            data[column] = None

    data = data[COLUMNS]

else:
    data = pd.DataFrame(columns=COLUMNS)


def choose(question, options):
    print(f"\n{question}")

    for number, option in enumerate(options, start=1):
        print(f"{number}. {option}")

    while True:
        try:
            answer = int(input("Enter your choice: "))

            if 1 <= answer <= len(options):
                return options[answer - 1]

            print("Choose one of the listed numbers.")

        except ValueError:
            print("Please enter a number.")


def rating(question, minimum, maximum):
    while True:
        try:
            answer = int(
                input(f"\n{question} ({minimum}-{maximum}): ")
            )

            if minimum <= answer <= maximum:
                return answer

            print(f"Enter a number from {minimum} to {maximum}.")

        except ValueError:
            print("Please enter a whole number.")


def collect_observation():
    global data

    print("\n=== BLUE NIKE HAT DATA COLLECTOR ===")
    print("Record a genuine occasion, not a hypothetical one.")

    hat_available = choose(
        "Was your blue Nike hat available?",
        ["Yes", "No"]
    )

    weather = choose(
        "What was the weather?",
        ["Sunny", "Cloudy", "Rainy", "Cold", "Other"]
    )

    activity = choose(
        "What were you doing?",
        [
            "Football",
            "Gym",
            "Going out",
            "Errands",
            "Travelling",
            "Other"
        ]
    )

    outfit_matches = choose(
        "Did your outfit match the hat?",
        ["Yes", "No", "Not applicable"]
    )

    mood = rating(
        "How strongly did your mood influence the decision?",
        1,
        5
    )

    if hat_available == "No":
        print(
            "\nBecause the hat was unavailable, "
            "you could not wear it."
        )

        wore_hat = 0

    else:
        wore_hat = 1 if choose(
            "Did you actually wear the blue Nike hat?",
            ["Yes", "No"]
        ) == "Yes" else 0

    record = {
        "date": datetime.now().strftime("%Y-%m-%d %H:%M"),
        "weather": weather,
        "hat_available": hat_available,
        "activity": activity,
        "outfit_matches": outfit_matches,
        "mood": mood,
        "nike_preference": 4,
        "wore_blue_nike_hat": wore_hat
    }

    data = pd.concat(
        [data, pd.DataFrame([record])],
        ignore_index=True
    )

    data.to_csv(FILE, index=False)

    print("\nObservation saved successfully!")
    print(f"Total recorded observations: {len(data)}")

    print("\nYour latest observation:")
    print(pd.DataFrame([record]).to_string(index=False))


collect_observation()
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ==========================================
# 1. LOAD YOUR DATASET
# ==========================================

FILE = "blue_nike_hat_data.csv"

data = pd.read_csv(FILE)

print("DATASET OVERVIEW")
print("================")
print("Total observations:", len(data))
print("\nFirst five observations:")
print(data.head())

# ==========================================
# 2. PREPARE THE DATA
# ==========================================

features = [
    "weather",
    "hat_available",
    "activity",
    "outfit_matches",
    "mood",
    "nike_preference"
]

target = "wore_blue_nike_hat"

# Ensure the required columns exist.
missing = set(features + [target]) - set(data.columns)

if missing:
    raise ValueError(f"Missing columns: {missing}")

# Remove incomplete records.
data = data.dropna(subset=features + [target]).copy()

# Convert the target to integer labels.
data[target] = pd.to_numeric(data[target], errors="raise").astype(int)

if not data[target].isin([0, 1]).all():
    raise ValueError("Target must contain only 0 and 1.")

X = data[features]
y = data[target]

print("\nOUTCOME DISTRIBUTION")
print("====================")
print(y.value_counts().sort_index())

# ==========================================
# 3. DEFINE DATA PREPROCESSING
# ==========================================

categorical_features = [
    "weather",
    "hat_available",
    "activity",
    "outfit_matches"
]

numerical_features = [
    "mood",
    "nike_preference"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="most_frequent")
                ),
                (
                    "encoder",
                    OneHotEncoder(handle_unknown="ignore")
                )
            ]),
            categorical_features
        ),
        (
            "numerical",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="median")
                )
            ]),
            numerical_features
        )
    ]
)

# ==========================================
# 4. CHECK WHETHER TRAINING IS POSSIBLE
# ==========================================

if len(data) < 12 or y.nunique() < 2:
    raise ValueError(
        "Collect more observations, including both outcomes."
    )

if y.value_counts().min() < 2:
    raise ValueError(
        "There are too few examples of one outcome."
    )

# ==========================================
# 5. SPLIT INTO TRAINING AND TEST DATA
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# ==========================================
# 6. BUILD AND TRAIN THE MODEL
# ==========================================

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)

print("\nMODEL TRAINED SUCCESSFULLY!")

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))

# ==========================================
# 7. EVALUATE THE MODEL
# ==========================================

predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)

print("\nMODEL EVALUATION")
print("================")
print(f"Test accuracy: {accuracy:.1%}")

print("\nClassification report:")
print(classification_report(
    y_test,
    predictions,
    labels=[0, 1],
    target_names=["Did not wear hat", "Wore hat"],
    zero_division=0
))

print("\nConfusion matrix:")
print(confusion_matrix(
    y_test,
    predictions,
    labels=[0, 1]
))

# ==========================================
# 8. PREDICT A NEW SITUATION
# ==========================================

new_situation = pd.DataFrame([{
    "weather": "Sunny",
    "hat_available": "Yes",
    "activity": "Football",
    "outfit_matches": "Yes",
    "mood": 4,
    "nike_preference": 4
}])

probabilities = model.predict_proba(new_situation)[0]

classes = model.named_steps["classifier"].classes_

probability_wear = probabilities[list(classes).index(1)]

print("\nNEW PREDICTION")
print("==============")
print(f"Probability of wearing the hat: {probability_wear:.1%}")

if probability_wear >= 0.5:
    print("Prediction: Likely to wear the hat.")
else:
    print("Prediction: Likely not to wear the hat.")

print(
    "\nIMPORTANT: This is an experimental prediction "
    "from a very small dataset, not a reliable estimate "
    "of your future behaviour."
)
```
